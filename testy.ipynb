```python
import pandas as pd
```

```python
df = pd.read_excel('data.xlsx', sheet_name='Sheet2')
```

```python
df.columns
```

```python
names = df['Gianluca'].str.strip().str.lower().to_list()
```

```python
names.append('gianluca')
```

```python
contador = {}

for nombre in names:
    for letra in nombre:
        if letra not in contador.keys():
            contador[letra] = 1
        elif letra in contador.keys():
            contador[letra] += 1
```

```python
contador
```

```python
df = pd.DataFrame(contador.items(), columns=['letras', 'valores'])
```

```python
len(names)
```

```python
df.sort_values(by='valores', ascending=False).to_excel('recuento_letras.xlsx', index=False)
```

```python
import pandas as pd
from google.cloud import bigquery
import pandas_gbq
from datetime import datetime, timedelta
```

```python
df = pd.read_excel('stock.xlsx')
df
```

```python
conn = bigquery.Client(project='hitobaby')
```

```python
datasets = conn.list_datasets()
```

```python
for dataset in datasets:
    print(dataset.dataset_id)
```

```python
df.columns
```

```python
df['codigo material'] = df['codigo material'].astype(str)
df['cantidad'] = df['cantidad'].astype(int)
df = df.rename(columns={'codigo material': 'codigo_material'})
df['fecha_modificacion'] = datetime.today().date()
```

```python
table_id = 'hitobaby.hitobaby_dataset.stock'
job_config = bigquery.LoadJobConfig(
    write_disposition='WRITE_TRUNCATE',
    autodetect=True
)
print('comenzando job ...')
job = conn.load_table_from_dataframe(df, table_id, job_config=job_config)

job.result()
print(f'Exito, se subio la tabla de forma correcta a Bigquery. Filas instertadas {df.shape[0]} a la tabla {table_id}')
```
